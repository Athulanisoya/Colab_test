# Severity dataset analysis

Inspect the 60 explicitly synthetic messages before training. These examples are not operational Kerala field data. All cells use the project's existing dependencies.

```python
from pathlib import Path
import sys
PROJECT_ROOT = next(path for path in (Path.cwd(), *Path.cwd().parents)
                    if (path / "ResQ_Kerala_Complete_Project_Discussion.md").is_file())
sys.path.insert(0, str(PROJECT_ROOT))
from backend.ai.severity_model import load_examples, new_pipeline, DISCLAIMER
rows = load_examples()
print(DISCLAIMER)
```

```python
from collections import Counter
import re
counts = Counter(row["severity"] for row in rows)
malayalam_count = sum(bool(re.search(r"[ഀ-ൿ]", row["message"])) for row in rows)
lengths = [len(row["message"]) for row in rows]
print({"messages": len(rows), "classes": dict(counts), "Malayalam_messages": malayalam_count,
       "min_characters": min(lengths), "max_characters": max(lengths),
       "duplicate_messages": len(rows) - len({row["message"] for row in rows})})
assert all(row["source"] == "synthetic_demo" for row in rows)
assert set(counts) == {"LOW", "MODERATE", "HIGH"}
```

The small balanced dataset supports a reproducible demonstration. Its counts do not establish real disaster prevalence or classifier reliability.
